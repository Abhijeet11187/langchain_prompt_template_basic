### Designing chatPromptTemplate using SystemMessagePromptTemplate, HumanMessagePromptTemplate and AIMessagePromptTemplate

```python
from langchain_core.prompts import ChatPromptTemplate,SystemMessagePromptTemplate,HumanMessagePromptTemplate,AIMessagePromptTemplate
```

```python
# System Template (Persona)

system_prompt_template=SystemMessagePromptTemplate.from_template(
    "You are a helpful AI bot ,your name is {bot_name}"
)

# Human Template

human_prompt_template=HumanMessagePromptTemplate.from_template(
    "Hello how are you dong today?"
)

# AI Template

ai_prompt_template = AIMessagePromptTemplate.from_template(
    "I am doing well, thanks"
)

# Human Template

human_prompt_template=HumanMessagePromptTemplate.from_template(
    "{user_input_prompt}"
)

# Compile a chat prompt

chat_template=ChatPromptTemplate.from_messages(
    [
        system_prompt_template,
        human_prompt_template,
        ai_prompt_template,
        human_prompt_template
    ]
)
```

```python
chat_template
```

```python
chat_template.input_variables
```

```python
chat_template.format_prompt(bot_name="Chitti",user_input_prompt="Tell me about yourself").to_messages()
```

```python
chat_template
```
